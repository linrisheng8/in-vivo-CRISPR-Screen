```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams  

rcParams['font.family'] = 'sans-serif'  
rcParams['font.sans-serif'] = ['Arial']  
rcParams['font.size'] = 10 
rcParams['axes.linewidth'] = 0.8  
rcParams['xtick.major.width'] = 0.8  
rcParams['ytick.major.width'] = 0.8  
rcParams['xtick.major.size'] = 3  
rcParams['ytick.major.size'] = 3  
rcParams['pdf.fonttype'] = 42  # Ensures text is editable in AI  
rcParams['ps.fonttype'] = 42  
rcParams['svg.fonttype'] = 'none'  
rcParams['figure.dpi'] = 300  
```

```python
df = pd.read_csv('../new_mageck/Vgat-ND-all_2M/Vgat-ND-all_2M.sgrna_summary.txt', sep = '\t')
df['log2 count-control'] = np.log2(df['control_mean'])
df['log2 count-treatment'] = np.log2(df['treat_mean'])
df_ctrl =df[df['Gene']=='non-targeting']
df_ost = df[df['Gene']=='Ostm1']
df_exo = df[df['Gene']=='Exosc9']
```

```python
# sns.scatterplot(df, x='control_mean', y='treat_mean')
plt.figure(figsize = [3,3])
ax = sns.scatterplot(df_ctrl, x='log2 count-control', y='log2 count-treatment', c = 'grey',
                    edgecolors='white',linewidths=0.1, s= 20, label = 'non-targeting')
ax = sns.scatterplot(df_exo, x='log2 count-control', y='log2 count-treatment', c = '#E64B35FF',
                    edgecolors='white',linewidths=0.1, label = 'Exosc9')
ax = sns.scatterplot(df_ost, x='log2 count-control', y='log2 count-treatment', c = 'purple',
                    edgecolors='white',linewidths=0.1, label = 'Ostm1')

ax.set_xlim(8,11.5)
ax.set_ylim(8,11.5)
plt.xticks(np.arange(8,12) ) # set x ticks at whole numbers  
plt.yticks(np.arange(8,12))  # set x ticks at whole numbers  
plt.legend()
plt.xlabel('log2 count-control')
plt.ylabel('log2 count-Vgat ND 2M')

plt.savefig('../ostm1_exosc9/vgat_ND_2M.pdf',bbox_inches = 'tight')
```
